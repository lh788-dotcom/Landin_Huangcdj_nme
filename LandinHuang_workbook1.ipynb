```python
#Libraries, loops, and lists
#import two standard libraries
import numpy as np
import pandas as pd
# you can use "as" to nickname to libraries
```

```python
#initialize an array
# make an array of numbers 0 to 100 (exclusive)

random_list = np.arange(stop = 100)

#consult documentation for further function parameters/outputs, etc...
```

```python
#print is built-in in python

print(random_list)
```

```python
#A1 use np.arange to create a new list
# called even_list_1 that contains all the even numbers from 0 to 100 (exclusive)

even_list_1 = np.arange(0, 100, 2).tolist()
print(even_list_1)
```

```python
#the [] allows us to access elements from an array 
print(random_list [0])

#use the : operator to access elements a certain range 
print(random_list[5:15])
#note that the last element is exclusive 

#we also just access a number at either end of your list using : to indicate we want Python to go all the way to end
print(random_list[95:])
print(random_list[:5])

```

```python
#A2

#print the element at the 90th index of the array 
ninety_index = random_list [90]
#print the elements from index 10 to index 21
index_ten_to_twenty_one = random_list [10:22]
#print the last ten elements of the array 
last_ten_indices = random_list[90:]


print(random_list [90])
print(random_list[10:22])
print(random_list[90:])
```

```python
#doing mathematical operations on our list
#want the list to start at 1 instead of 0 
#add 1 to each element in the array 

random_list = random_list + 1
print(random_list)
```

```python
#do the same with natural nlog, using np.log command
#take the natural log of each element in the array, and printing the result

print(np.log(random_list))
```

```python
#boolean indexing can also be performed element-wise across an array
#boolean index array that shwos where all the odds(falses) are:

print(random_list % 2 == 0 )

#taking in the boolean index array into the new array allows us to filter out the falses 
even_list_2 = random_list[random_list % 2 == 0]
print(even_list_2)
```

```python
#initialize an empty array 
even_list_3 = np.array([])

#for i in range just means for each value (i) in the range from 0 and 101 (exclusive)
for i in range (0,101):
    #we check if the value (i) is even (when divided by 2, remainder = 0 )
    if i % 2 == 0:
        #use the np.append function to apend i since it's even
        even_list_3 = np.append(arr = even_list_3, values = i)
        #notice the append function takes in an array named 'arr' and the value you'd like to append


print(even_list_3)
```

```python
#make an array of number from 0 to 200 (exclusive)
big_list = np.arange(200)

#multiply numbers resulting in a big list
even_list_4 = [2*x for x in big_list if x <= 50] 
#notice how list comprehsnion allows us to save lines of code
#perform multiple operations

#convert to an array 
even_list_4 = np.array(even_list_4)

print(even_list_4)
```

```python
#II. Dataframes and Pandas

gapminder = pd.read_csv('gapminder.csv')
```
